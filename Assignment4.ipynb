```python
import unified_planning as up
from unified_planning.io import PDDLReader
from unified_planning.engines import OperationMode
from unified_planning.shortcuts import get_all_applicable_engines

# Disable credits
up.shortcuts.get_environment().credits_stream = None

# Parse your PDDL files
reader     = PDDLReader()
up_problem = reader.parse_problem("domain.pddl", "problem.pddl")

# List all engines that can handle One-shot planning
available = get_all_applicable_engines(
    up_problem.kind,
    operation_mode=OperationMode.ONESHOT_PLANNER
)
print("Registered OneshotPlanner engines:", available)
```

```python
import unified_planning as up
from unified_planning.io import PDDLReader
from unified_planning.shortcuts import OneshotPlanner
from unified_planning.engines.results import POSITIVE_OUTCOMES

# 1) Disable planner credits printing
up.shortcuts.get_environment().credits_stream = None

# 2) Load domain + problem
reader     = PDDLReader()
up_problem = reader.parse_problem("domain.pddl", "problem.pddl")

# 3) Solve with Pyperplan
with OneshotPlanner(name="pyperplan", problem_kind=up_problem.kind) as planner:
    result = planner.solve(up_problem)

# 4) Check status against POSITIVE_OUTCOMES
if result.status in POSITIVE_OUTCOMES:
    print("Plan found:")
    for step in result.plan.actions:
        print("  ", step)
else:
    print("No plan found (status:", result.status, ")")
```
